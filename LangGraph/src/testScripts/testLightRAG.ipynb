```python
from __future__ import annotations
import os
import asyncio
from typing import Optional, Dict, Any, List
from dotenv import load_dotenv
from lightrag import LightRAG, QueryParam
from lightrag.llm.openai import openai_complete_if_cache, openai_embed
from lightrag.kg.shared_storage import initialize_pipeline_status
from lightrag.utils import setup_logger
import logging

# Load environment variables
load_dotenv("../../../.env.lightrag")

# Setup logger
setup_logger("lightrag", level=os.getenv("LIGHTRAG_LOG_LEVEL", "INFO"))
logger = logging.getLogger("lightrag_core")
```

```python
async def initialize_rag():
    rag = LightRAG(
        self,
        working_dir=Optional[str],
        llm_model=Optional[str],
        llm_api_base=Optional[str],
        llm_api_key=Optional[str],
        embedding_model=Optional[str],
        embedding_api_base=Optional[str],
        embedding_dim=Optional[int],
        kv_storage=Optional[str],
        vector_storage=Optional[str],
        graph_storage=Optional[str],
        doc_status_storage=Optional[str]
    )
    # IMPORTANT=Both initialization calls are required!
    await rag.initialize_storages()  # Initialize storage backends
    await initialize_pipeline_status()  # Initialize processing pipeline
    return rag
```
