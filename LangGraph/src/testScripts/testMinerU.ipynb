```python
import os
import sys
from dotenv import load_dotenv
load_dotenv()
project_dir = str(os.environ.get("PROJECT_ROOT"))

src_path = os.path.join(project_dir, "LangGraph", "src")
print(src_path)
sys.path.append(os.path.abspath(src_path))
```

```python
from pathlib import Path
from agent.mineru_client import MinerUClient, MinerUClientError
mineru_client = MinerUClient()
```

```python
file_path = "/Users/jajajou1778/UIT_DOCS_AGENT/firecrawl/data/daa/quydinh_huongdan/huong-dan-chuan-qua-trinh/pdf/547-qd-dhcntt_30-8-2019_qui_dinh_dao_tao_ngoai_ngu_doi_voi_he_chinh_qui_khoa_2019_0_0.pdf"

if not os.path.exists(file_path):
    raise FileNotFoundError(f"File not found: {file_path}")

file_stem = Path(file_path).stem
output_dir = f"./output/{file_stem}"
os.makedirs(output_dir, exist_ok=True)

print(file_stem)
print(output_dir)
```

```python
try:
    print(f"[MINERU] Parsing {os.path.basename(file_path)} ...")

    md_content, output_path = mineru_client.parse_and_get_markdown(
        file_path=file_path,
        output_dir=output_dir,
        parse_method="auto",
        lang_list="vi",  
        table_enable=True,
        formula_enable=True,
    )

    print(f"[✓] Parsed successfully — Markdown length: {len(md_content):,} chars")
    print(f"Output directory: {output_path}")

    # Lưu kết quả markdown để xem lại
    md_out = os.path.join(output_dir, f"{file_stem}.md")
    with open(md_out, "w", encoding="utf-8") as f:
        f.write(md_content)
    print(f"[💾] Markdown saved to {md_out}")

except MinerUClientError as e:
    print(f"[MinerUClientError] {e}")
except Exception as e:
    print(f"[❌] Unexpected error: {e}")
```
