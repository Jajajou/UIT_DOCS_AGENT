```python
import os
from pathlib import Path
from mineru_client import MinerUClient, MinerUClientError
mineru_client = MinerUClient()
```

```python
file_path = "/Users/jajajou1778/UIT_DOCS_AGENT/firecrawl/data/daa/quydinh_huongdan/huong-dan-chuan-qua-trinh/pdf/03-tb-dhcntt_17-1-2017.pdf"

if not os.path.exists(file_path):
    raise FileNotFoundError(f"File not found: {file_path}")

file_stem = Path(file_path).stem
output_dir = f"./output/{file_stem}"
os.makedirs(output_dir, exist_ok=True)
```

```python
try:
    print(f"[MINERU] Parsing {os.path.basename(file_path)} ...")

    md_content, output_path = mineru_client.parse_and_get_markdown(
        file_path=file_path,
        output_dir=output_dir,
        parse_method="auto",
        lang_list="vi",   # ✅ sửa: "en" thay cho "latin"
        table_enable=True,
        formula_enable=True,
    )

    print(f"[✓] Parsed successfully — Markdown length: {len(md_content):,} chars")
    print(f"Output directory: {output_path}")

    # Lưu kết quả markdown để xem lại
    md_out = os.path.join(output_dir, f"{file_stem}.md")
    with open(md_out, "w", encoding="utf-8") as f:
        f.write(md_content)
    print(f"[💾] Markdown saved to {md_out}")

except MinerUClientError as e:
    print(f"[MinerUClientError] {e}")
except Exception as e:
    print(f"[❌] Unexpected error: {e}")
```
