```python
def read_markdown_file(file_path):
    """
    Reads the content of a Markdown file into a single string.

    Args:
        file_path (str): The path to the Markdown file.

    Returns:
        str: The full content of the file as text.
    """
    try:
        # Use 'utf-8' encoding for general compatibility with various text files
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()
        return content
    except FileNotFoundError:
        return f"Error: The file at '{file_path}' was not found."
    except IOError as e:
        return f"Error reading file: {e}"

# Example Usage:
# Assume you have a file named 'README.md' in the same directory
file_name = '/Users/jajajou1778/UIT_DOCS_AGENT/data/DeepSeek-OCR/547-qd-dhcntt_30-8-2019_qui_dinh_dao_tao_ngoai_ngu_doi_voi_he_chinh_qui_khoa_2019_0_0/547-qd-dhcntt_30-8-2019_qui_dinh_dao_tao_ngoai_ngu_doi_voi_he_chinh_qui_khoa_2019_0_0.md'
markdown_text = read_markdown_file(file_name)

if not markdown_text.startswith("Error"):
    print("File content successfully read:")
    print("-" * 20)
    print(markdown_text)
else:
    print(markdown_text)

```

```python
ollama_client.embed(
    model='hf.co/doof-ferb/AITeamVN-Vietnamese_Embedding-gguf:F32',
    input=markdown_text
)
```
